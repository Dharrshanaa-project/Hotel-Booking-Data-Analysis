```python
df["estimated_revenue"] = (
    df["adr"] * df["total_stay_nights"]
)

print(df[
    ["adr", "total_stay_nights", "estimated_revenue"]
].head())
```

```python
total_revenue = df["estimated_revenue"].sum()

print(
    "Total Estimated Revenue:",
    round(total_revenue, 2)
)
```

```python
hotel_revenue = (
    df.groupby("hotel")["estimated_revenue"]
    .sum()
)

print("Estimated Revenue by Hotel:")
print(hotel_revenue)
```

```python
plt.figure(figsize=(8, 5))

sns.barplot(
    x=hotel_revenue.index,
    y=hotel_revenue.values
)

plt.title("Estimated Revenue by Hotel Type")
plt.xlabel("Hotel Type")
plt.ylabel("Estimated Revenue")

plt.show()
```

```python
# STEP 12 - Final Project Summary

# 1. Total bookings
total_bookings = len(df)

# 2. Cancelled bookings
cancelled_bookings = df["is_canceled"].sum()

# 3. Cancellation rate
cancellation_rate = (
    cancelled_bookings / total_bookings
) * 100

# 4. Most booked hotel
most_booked_hotel = df["hotel"].value_counts().idxmax()

# 5. Most common customer type
most_common_customer = df["customer_type"].value_counts().idxmax()

# 6. Most common country
most_common_country = df["country"].value_counts().idxmax()

# 7. Most reserved room
most_reserved_room = df["reserved_room_type"].value_counts().idxmax()

# 8. Average stay
average_stay = df["total_stay_nights"].mean()

# 9. Average ADR
average_adr = df["adr"].mean()

# 10. Estimated revenue
total_revenue = df["estimated_revenue"].sum()

print("===== HOTEL BOOKING ANALYSIS SUMMARY =====")

print("Total Bookings:", total_bookings)
print("Cancelled Bookings:", cancelled_bookings)
print("Cancellation Rate:", round(cancellation_rate, 2), "%")
print("Most Booked Hotel:", most_booked_hotel)
print("Most Common Customer Type:", most_common_customer)
print("Most Common Country:", most_common_country)
print("Most Reserved Room Type:", most_reserved_room)
print("Average Stay:", round(average_stay, 2), "nights")
print("Average ADR:", round(average_adr, 2))
print("Estimated Revenue:", round(total_revenue, 2))
```
